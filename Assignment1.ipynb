## Question 1

Given $P(A)=0.4$ and $P(B)=0.3$.

Since $A$ and $B$ are independent,

$$
P(A \cap B)=P(A)P(B)=0.4\times0.3=0.12
$$

For the union,

$$
P(A \cup B)=P(A)+P(B)-P(A \cap B)
$$

$$
=0.4+0.3-0.12=0.58
$$

**Answer:** $P(A \cap B)=0.12$ and $P(A \cup B)=0.58$.

## Question 2

Given $P(A)=0.5$, $P(B)=0.4$, and $P(A|B)=0.7$.

If $A$ and $B$ were independent, then

$$
P(A|B)=P(A)
$$

But here,

$$
0.7 \neq 0.5
$$

so $A$ and $B$ are **not independent**.

We can also check:

$$
P(A \cap B)=P(A|B)P(B)=0.7\times0.4=0.28
$$

If they were independent,

$$
P(A \cap B)=P(A)P(B)=0.5\times0.4=0.20
$$

Since $0.28 \neq 0.20$, the events are not independent.

## Question 3

Given $P(A)=0.6$, $P(B|A)=0.5$, and $P(B|A')=0.2$.

First,

$$
P(A')=1-0.6=0.4
$$

Using the law of total probability,

$$
P(B)=P(B|A)P(A)+P(B|A')P(A')
$$

$$
=(0.5)(0.6)+(0.2)(0.4)=0.30+0.08=0.38
$$

Using Bayes' Rule,

$$
P(A|B)=\frac{P(B|A)P(A)}{P(B)}
$$

$$
=\frac{(0.5)(0.6)}{0.38}
=\frac{0.30}{0.38}
\approx 0.7895
$$

So,

$$
P(A|B)\approx 78.95\%
$$

## Question 4

Let $D$ mean the person has the disease and $+$ mean a positive test.

Given:

$$
P(D)=0.02, \quad P(+|D)=0.95
$$

The true negative rate is 90%, so the false positive rate is

$$
P(+|D')=1-0.90=0.10
$$

Also,

$$
P(D')=1-0.02=0.98
$$

The total probability of a positive test is

$$
P(+)=P(+|D)P(D)+P(+|D')P(D')
$$

$$
=(0.95)(0.02)+(0.10)(0.98)
$$

$$
=0.019+0.098=0.117
$$

Using Bayes' Rule,

$$
P(D|+)=\frac{P(+|D)P(D)}{P(+)}
$$

$$
=\frac{0.019}{0.117}\approx0.1624
$$

So the probability that a person who tests positive actually has the disease is about **16.24%**.

## Question 5

The distribution is:

| X | 85 | 90 | 95 | 100 |
|---|---:|---:|---:|---:|
| P(X=x) | 0.375 | 0.375 | 0.125 | 0.125 |

### (a) Expected value

$$
E[X]=\sum xP(X=x)
$$

$$
=(85)(0.375)+(90)(0.375)+(95)(0.125)+(100)(0.125)
$$

$$
E[X]=90
$$

### (b) Variance

First,

$$
E[X^2]=(85^2)(0.375)+(90^2)(0.375)+(95^2)(0.125)+(100^2)(0.125)
$$

$$
E[X^2]=8125
$$

Then,

$$
Var(X)=E[X^2]-(E[X])^2
$$

$$
=8125-90^2=25
$$

### (c) Sample mean

For the sample $\{85,90,85,95,90,85,100,90\}$,

$$
\bar{x}=\frac{85+90+85+95+90+85+100+90}{8}
$$

$$
\bar{x}=\frac{720}{8}=90
$$

The sample mean is the same as the expected value.

```python
scores = [85, 90, 95, 100]
probs = [0.375, 0.375, 0.125, 0.125]

expected_value = sum(x * p for x, p in zip(scores, probs))
expected_x2 = sum((x ** 2) * p for x, p in zip(scores, probs))
variance = expected_x2 - expected_value ** 2

sample = [85, 90, 85, 95, 90, 85, 100, 90]
sample_mean = sum(sample) / len(sample)

expected_value, variance, sample_mean
```

## Question 6

The message probabilities are:

| Message | A | B | C | D |
|---|---:|---:|---:|---:|
| Probability | 0.4 | 0.3 | 0.2 | 0.1 |

### (a) Entropy

$$
H(X)=-\sum P(X=x)\log_2 P(X=x)
$$

$$
H(X)=-
\left[
0.4\log_2(0.4)+0.3\log_2(0.3)+0.2\log_2(0.2)+0.1\log_2(0.1)
\right]
$$

$$
H(X)\approx1.8464\text{ bits}
$$

### (b) Equal probabilities

If all four messages have probability 0.25,

$$
H(X)=-4(0.25\log_2 0.25)
$$

Since $\log_2(0.25)=-2$,

$$
H(X)=2\text{ bits}
$$

```python
import math

probs = [0.4, 0.3, 0.2, 0.1]
entropy = -sum(p * math.log2(p) for p in probs)

equal_probs = [0.25, 0.25, 0.25, 0.25]
equal_entropy = -sum(p * math.log2(p) for p in equal_probs)

entropy, equal_entropy
```
